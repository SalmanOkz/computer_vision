```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load a noisy grayscale image
img = cv2.imread('/content/download.jpg', cv2.IMREAD_GRAYSCALE)

# Optional: Simulate salt-and-pepper noise if not already noisy
def add_salt_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    noisy = np.copy(image)
    total_pixels = image.size
    num_salt = np.ceil(salt_prob * total_pixels)
    num_pepper = np.ceil(pepper_prob * total_pixels)

    # Add salt noise (white pixels)
    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in image.shape]
    noisy[coords[0], coords[1]] = 255

    # Add pepper noise (black pixels)
    coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in image.shape]
    noisy[coords[0], coords[1]] = 0
    return noisy

noisy_img = add_salt_pepper_noise(img)

# Step 2: Apply filters (kernel size = 5x5)
mean_filtered = cv2.blur(noisy_img, (5,5))
gaussian_filtered = cv2.GaussianBlur(noisy_img, (5,5), 0)
median_filtered = cv2.medianBlur(noisy_img, 5)

# Step 3: Manual Histogram Equalization
def manual_hist_equalization(image):
    hist, bins = np.histogram(image.flatten(), 256, [0,256])
    cdf = hist.cumsum()
    cdf_normalized = cdf * hist.max()/ cdf.max()

    cdf_m = np.ma.masked_equal(cdf, 0)
    cdf_m = (cdf_m - cdf_m.min())*255/(cdf_m.max()-cdf_m.min())
    cdf = np.ma.filled(cdf_m, 0).astype('uint8')
    equalized_image = cdf[image]
    return equalized_image

equalized_img = manual_hist_equalization(median_filtered)

# Step 4: Display all images
titles = ['Original Noisy', 'Mean Filter', 'Gaussian Filter', 'Median Filter', 'Histogram Equalized']
images = [noisy_img, mean_filtered, gaussian_filtered, median_filtered, equalized_img]

plt.figure(figsize=(15,8))
for i in range(5):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load a grayscale image
img = cv2.imread('/content/download.jpg', cv2.IMREAD_GRAYSCALE)

# Step 2: Define the manual mean filter function (3x3)
def mean_filter_manual(image, kernel_size=3):
    # Add padding to the image
    padded_image = np.pad(image, pad_width=kernel_size//2, mode='constant', constant_values=0)

    # Get image dimensions
    height, width = image.shape

    # Create an empty output image
    output_image = np.zeros_like(image)

    # Apply the 3x3 mean filter using nested loops
    for i in range(height):
        for j in range(width):
            # Extract the 3x3 region around the current pixel
            region = padded_image[i:i+kernel_size, j:j+kernel_size]

            # Compute the mean of the region
            output_image[i, j] = np.mean(region)

    return output_image

# Step 3: Apply manual mean filter
manual_filtered = mean_filter_manual(img)

# Step 4: Apply OpenCV mean filter (cv2.blur)
opencv_filtered = cv2.blur(img, (3, 3))

# Step 5: Display both images for comparison
titles = ['Original Image', 'Manual Mean Filter', 'OpenCV Mean Filter']
images = [img, manual_filtered, opencv_filtered]

plt.figure(figsize=(12,6))
for i in range(3):
    plt.subplot(1, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load a low-quality medical image
img = cv2.imread('/content/x-ray.jpg', 0)  # Read as grayscale

# If no image file, create a sample low-contrast image
if img is None:
    print("Creating sample low-contrast image...")
    img = np.zeros((200, 300), dtype=np.uint8)
    # Create gradient
    for i in range(200):
        for j in range(300):
            img[i, j] = min(j // 3, 100)  # Low contrast range

# Show original image
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Manual Histogram Equalization
height, width = img.shape
total_pixels = height * width

# Step 1: Compute histogram
histogram = np.zeros(256)
for i in range(height):
    for j in range(width):
        histogram[img[i, j]] += 1

# Step 2: Normalize (convert to probability)
normalized_hist = histogram / total_pixels

# Step 3: Find cumulative distribution (CDF)
cdf = np.zeros(256)
cdf[0] = normalized_hist[0]
for i in range(1, 256):
    cdf[i] = cdf[i-1] + normalized_hist[i]

# Step 4: Map old intensity values to new equalized values
equalized_img = np.zeros_like(img)
for i in range(height):
    for j in range(width):
        old_val = img[i, j]
        new_val = int(cdf[old_val] * 255)
        equalized_img[i, j] = new_val

# Show equalized image
plt.subplot(1, 2, 2)
plt.imshow(equalized_img, cmap='gray')
plt.title('Manually Equalized Image')
plt.axis('off')
plt.tight_layout()
plt.show()

# Show histograms comparison
plt.figure(figsize=(12, 4))

# Original histogram
plt.subplot(1, 3, 1)
plt.bar(range(256), histogram)
plt.title('Original Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')

# CDF
plt.subplot(1, 3, 2)
plt.plot(range(256), cdf * total_pixels)  # Now total_pixels is defined
plt.title('Cumulative Distribution (CDF)')
plt.xlabel('Intensity')
plt.ylabel('Cumulative Frequency')

# Equalized histogram
equalized_hist = np.zeros(256)
for i in range(height):
    for j in range(width):
        equalized_hist[equalized_img[i, j]] += 1

plt.subplot(1, 3, 3)
plt.bar(range(256), equalized_hist)
plt.title('Equalized Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

print(f"Image size: {height}x{width}")
print(f"Total pixels: {total_pixels}")
print("\n✅ Histogram equalization completed successfully!")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load a low-contrast medical X-ray image
img = cv2.imread('/content/x-ray.jpg', 0)  # 0 = grayscale mode

# If image not found, create a sample low-contrast image
if img is None:
    print("Image not found. Creating sample low-contrast X-ray...")
    # Create a sample low-contrast image (typical X-ray intensity range)
    img = np.zeros((400, 500), dtype=np.uint8)
    height, width = img.shape
    for i in range(height):
        for j in range(width):
            # Simulate X-ray with limited contrast (range 60-150 instead of 0-255)
            base_intensity = 100
            variation = 40 * np.sin(i/50) * np.cos(j/50)
            img[i, j] = np.clip(base_intensity + variation, 60, 150)

# Display original image
print(f"Original image shape: {img.shape}")
print(f"Original pixel range: [{img.min()}, {img.max()}]")

# 1. Manual Linear Contrast Stretching
def linear_contrast_stretch(image):
    # Find r_min and r_max
    r_min = np.min(image)
    r_max = np.max(image)

    print(f"\nContrast Stretching Parameters:")
    print(f"Minimum intensity (r_min): {r_min}")
    print(f"Maximum intensity (r_max): {r_max}")
    print(f"Original range width: {r_max - r_min}")

    # Handle case where all pixels have same value
    if r_max == r_min:
        print("Warning: All pixels have same value. Cannot stretch contrast.")
        return image.copy()

    # Apply linear contrast stretching formula
    stretched = np.zeros_like(image, dtype=np.float32)

    # Using the formula: s = (r - r_min) * 255 / (r_max - r_min)
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            r = image[i, j]
            stretched[i, j] = ((r - r_min) * 255.0) / (r_max - r_min)

    # Convert back to uint8
    stretched = np.clip(stretched, 0, 255).astype(np.uint8)

    print(f"Stretched range: [{stretched.min()}, {stretched.max()}]")
    print(f"Contrast improvement: {(stretched.max() - stretched.min()) / (r_max - r_min):.2f}x")

    return stretched

# Apply contrast stretching
stretched_img = linear_contrast_stretch(img)

# 2. Using OpenCV's equalization for comparison
equalized_img = cv2.equalizeHist(img)

# Display all results
plt.figure(figsize=(15, 10))

# Original image
plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title(f'Original X-ray\nRange: [{img.min()}, {img.max()}]')
plt.colorbar()
plt.axis('off')

# Contrast stretched image
plt.subplot(2, 3, 2)
plt.imshow(stretched_img, cmap='gray', vmin=0, vmax=255)
plt.title(f'Contrast Stretched\nRange: [{stretched_img.min()}, {stretched_img.max()}]')
plt.colorbar()
plt.axis('off')

# Histogram equalized image
plt.subplot(2, 3, 3)
plt.imshow(equalized_img, cmap='gray', vmin=0, vmax=255)
plt.title(f'Histogram Equalized\nRange: [{equalized_img.min()}, {equalized_img.max()}]')
plt.colorbar()
plt.axis('off')

# Original histogram
plt.subplot(2, 3, 4)
plt.hist(img.ravel(), bins=256, range=(0, 256), color='blue', alpha=0.7)
plt.title('Original Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Contrast stretched histogram
plt.subplot(2, 3, 5)
plt.hist(stretched_img.ravel(), bins=256, range=(0, 256), color='green', alpha=0.7)
plt.title('Contrast Stretched Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Histogram equalized histogram
plt.subplot(2, 3, 6)
plt.hist(equalized_img.ravel(), bins=256, range=(0, 256), color='red', alpha=0.7)
plt.title('Histogram Equalized Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Quantitative comparison
print("\n" + "="*60)
print("QUANTITATIVE COMPARISON")
print("="*60)

# Calculate contrast metrics
def calculate_contrast_metrics(image, name):
    mean_intensity = np.mean(image)
    std_intensity = np.std(image)  # Standard deviation as contrast measure
    intensity_range = image.max() - image.min()

    print(f"\n{name}:")
    print(f"  Mean intensity: {mean_intensity:.2f}")
    print(f"  Standard deviation: {std_intensity:.2f} (higher = more contrast)")
    print(f"  Intensity range: {intensity_range}")

    return std_intensity

orig_std = calculate_contrast_metrics(img, "Original Image")
stretched_std = calculate_contrast_metrics(stretched_img, "Contrast Stretched")
equalized_std = calculate_contrast_metrics(equalized_img, "Histogram Equalized")

print("\n" + "="*60)
print("KEY OBSERVATIONS:")
print("="*60)
print("1. Contrast Stretching:")
print("   - Preserves relative brightness distribution")
print("   - Linearly expands intensity range to [0, 255]")
print(f"   - Contrast improvement: {stretched_std/orig_std:.2f}x")

print("\n2. Histogram Equalization:")
print("   - More aggressive contrast enhancement")
print("   - Can cause artifacts in homogeneous regions")
print("   - May over-enhance noise")
print(f"   - Contrast improvement: {equalized_std/orig_std:.2f}x")

print(f"\n3. Brightness preservation:")
brightness_change_stretch = abs(np.mean(stretched_img) - np.mean(img))
brightness_change_equalize = abs(np.mean(equalized_img) - np.mean(img))
print(f"   - Contrast stretching brightness change: {brightness_change_stretch:.2f}")
print(f"   - Histogram equalization brightness change: {brightness_change_equalize:.2f}")

if brightness_change_stretch < brightness_change_equalize:
    print("   ✓ Contrast stretching better preserves original brightness")
else:
    print("   ⚠️ Histogram equalization changed brightness more significantly")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read an image in grayscale (medical image example)
img = cv2.imread('/content/x-ray.jpg', 0)  # 0 = grayscale

# If image not found, create a synthetic medical image
if img is None:
    print("Image not found. Creating synthetic medical image...")
    # Create a synthetic image with varying contrast regions (like an X-ray)
    img = np.zeros((400, 500), dtype=np.uint8)

    # Add different regions with varying contrast
    # Region 1: Low contrast area (simulating soft tissue)
    img[100:200, 50:200] = np.random.normal(80, 5, (100, 150)).clip(70, 90).astype(np.uint8)

    # Region 2: Medium contrast area (simulating bone)
    img[50:150, 300:450] = np.random.normal(150, 15, (100, 150)).clip(130, 170).astype(np.uint8)

    # Region 3: High contrast area (simulating air/tissue boundary)
    img[250:350, 100:300] = np.random.normal(40, 20, (100, 200)).clip(20, 100).astype(np.uint8)

    # Add some gradient background
    for i in range(400):
        for j in range(500):
            img[i, j] = np.clip(img[i, j] + j//10 - i//20, 0, 255)

# Display original image
print(f"Original image shape: {img.shape}")
print(f"Original intensity range: [{img.min()}, {img.max()}]")

# 1. Global Histogram Equalization (for comparison)
global_eq = cv2.equalizeHist(img)

# 2. Create CLAHE object and apply
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_img = clahe.apply(img)

# Try different parameters for comparison
clahe_low_clip = cv2.createCLAHE(clipLimit=1.0, tileGridSize=(8, 8))
clahe_img_low = clahe_low_clip.apply(img)

clahe_high_clip = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
clahe_img_high = clahe_high_clip.apply(img)

# Display results
plt.figure(figsize=(16, 12))

# Original image
plt.subplot(3, 4, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Medical Image')
plt.axis('off')
plt.colorbar()

# Global histogram equalization
plt.subplot(3, 4, 2)
plt.imshow(global_eq, cmap='gray')
plt.title('Global Histogram Equalization')
plt.axis('off')
plt.colorbar()

# CLAHE enhanced
plt.subplot(3, 4, 3)
plt.imshow(clahe_img, cmap='gray')
plt.title('CLAHE Enhanced\n(clipLimit=2.0, grid=8x8)')
plt.axis('off')
plt.colorbar()

# CLAHE with low clip limit
plt.subplot(3, 4, 4)
plt.imshow(clahe_img_low, cmap='gray')
plt.title('CLAHE (clipLimit=1.0)')
plt.axis('off')
plt.colorbar()

# Original histogram
plt.subplot(3, 4, 5)
plt.hist(img.ravel(), bins=256, range=(0, 256), color='blue', alpha=0.7)
plt.title('Original Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Global equalization histogram
plt.subplot(3, 4, 6)
plt.hist(global_eq.ravel(), bins=256, range=(0, 256), color='red', alpha=0.7)
plt.title('Global Equalization Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# CLAHE histogram
plt.subplot(3, 4, 7)
plt.hist(clahe_img.ravel(), bins=256, range=(0, 256), color='green', alpha=0.7)
plt.title('CLAHE Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# CLAHE low clip histogram
plt.subplot(3, 4, 8)
plt.hist(clahe_img_low.ravel(), bins=256, range=(0, 256), color='purple', alpha=0.7)
plt.title('CLAHE (clip=1.0) Histogram')
plt.xlabel('Intensity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Intensity profiles across a line (to show local enhancement)
plt.subplot(3, 4, 9)
line_y = 200  # Horizontal line position
plt.plot(img[line_y, :], label='Original', linewidth=2)
plt.plot(global_eq[line_y, :], label='Global EQ', linewidth=1.5)
plt.plot(clahe_img[line_y, :], label='CLAHE', linewidth=1.5)
plt.title(f'Intensity Profile (Row {line_y})')
plt.xlabel('Column')
plt.ylabel('Intensity')
plt.legend()
plt.grid(True, alpha=0.3)

# Zoomed region comparison
zoom_region = (100, 200, 150, 250)  # y1, y2, x1, x2
y1, y2, x1, x2 = zoom_region

plt.subplot(3, 4, 10)
plt.imshow(img[y1:y2, x1:x2], cmap='gray')
plt.title('Original (Zoomed)')
plt.axis('off')

plt.subplot(3, 4, 11)
plt.imshow(global_eq[y1:y2, x1:x2], cmap='gray')
plt.title('Global EQ (Zoomed)')
plt.axis('off')

plt.subplot(3, 4, 12)
plt.imshow(clahe_img[y1:y2, x1:x2], cmap='gray')
plt.title('CLAHE (Zoomed)')
plt.axis('off')

plt.tight_layout()
plt.show()

# Quantitative analysis
print("\n" + "="*60)
print("QUANTITATIVE ANALYSIS")
print("="*60)

def calculate_image_stats(image, name):
    """Calculate image statistics"""
    mean_val = np.mean(image)
    std_val = np.std(image)  # Standard deviation = contrast measure
    min_val = np.min(image)
    max_val = np.max(image)
    range_val = max_val - min_val

    print(f"\n{name}:")
    print(f"  Mean intensity: {mean_val:.2f}")
    print(f"  Standard deviation: {std_val:.2f} (contrast measure)")
    print(f"  Intensity range: [{min_val}, {max_val}] = {range_val}")

    return std_val

# Calculate statistics
orig_std = calculate_image_stats(img, "Original Image")
global_std = calculate_image_stats(global_eq, "Global Equalization")
clahe_std = calculate_image_stats(clahe_img, "CLAHE Enhanced")
clahe_low_std = calculate_image_stats(clahe_img_low, "CLAHE (clip=1.0)")

print("\n" + "="*60)
print("PERFORMANCE COMPARISON")
print("="*60)
print(f"Contrast improvement (std dev ratio):")
print(f"  Global EQ / Original: {global_std/orig_std:.2f}x")
print(f"  CLAHE / Original: {clahe_std/orig_std:.2f}x")
print(f"  CLAHE (clip=1.0) / Original: {clahe_low_std/orig_std:.2f}x")

# Show CLAHE parameters explanation
print("\n" + "="*60)
print("CLAHE PARAMETERS EXPLANATION")
print("="*60)
print("1. clipLimit: Contrast limiting threshold (default: 2.0)")
print("   - Lower values = less contrast enhancement")
print("   - Higher values = more aggressive enhancement")
print("   - Prevents over-amplification of noise")

print("\n2. tileGridSize: Number of tiles in each dimension (default: 8x8)")
print("   - Smaller tiles = more local adaptation")
print("   - Larger tiles = more global behavior")
print("   - Typical values: 4x4 to 16x16")

# Interactive parameter exploration
def apply_clahe_with_params(image, clip_limit, grid_size):
    """Apply CLAHE with custom parameters"""
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=(grid_size, grid_size))
    return clahe.apply(image)

# Example: Try different grid sizes
print("\n" + "="*60)
print("TRY DIFFERENT PARAMETERS")
print("="*60)
print("Example: Different grid sizes for CLAHE:")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
grid_sizes = [4, 8, 16]

for idx, grid_size in enumerate(grid_sizes):
    clahe_custom = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(grid_size, grid_size))
    result = clahe_custom.apply(img)

    axes[idx].imshow(result, cmap='gray')
    axes[idx].set_title(f'CLAHE (grid={grid_size}x{grid_size})')
    axes[idx].axis('off')

    # Calculate stats
    std_val = np.std(result)
    print(f"  Grid {grid_size}x{grid_size}: std dev = {std_val:.2f}")

plt.tight_layout()
plt.show()

print("\n✅ CLAHE successfully applied! Key observations:")
print("   - CLAHE enhances local contrast without over-amplifying noise")
print("   - Better detail preservation than global histogram equalization")
print("   - Particularly useful for medical images with varying tissue densities")
```
